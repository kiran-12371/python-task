```python
#Personal Expense Tracker
total_expense = int(input("How many expenses do you want to add: "))

expenses_list = []

for i in range(1, total_expense + 1):
  name = input("Enter your expense name: ").title()
  category = input("Enter your expense category: ").title()
  amount = float(input("Enter your expense amount: "))

  expense_item = {
      "expense_name" : name,
      "category" : category,
      "amount": amount
  }

  expenses_list.append(expense_item)

total_spending = 0
for expense_item in expenses_list:
  total_spending = total_spending + expense_item["amount"]

category_spending = {
    "Food": 0,
    "Transport": 0,
    "Shopping": 0,
    "Bills": 0,
    "Other": 0
}

for expense_item in expenses_list:
    category = expense_item["category"]
    amount = expense_item["amount"]

    if category in category_spending:
        category_spending[category] = category_spending[category] + amount
    else:
        category_spending["Other"] = category_spending["Other"] + amount

highest_expense = expenses_list[0]
for expense_item in expenses_list:

    if expense_item["amount"] > highest_expense["amount"]:
        highest_expense = expense_item

budget = float(input("Enter your budget: "))

if total_spending > budget:
    print("Warning! Budget exceeded.")
else:
    print("You are within your budget.")

def final_report():

    print("\n----- Expense Report -----")

    print("Total Spending:", total_spending)

    print("\nCategory Spending")

    for category, amount in category_spending.items():
        print(category, ":", amount)

    print("\nHighest Expense")
    print("Name:", highest_expense["expense_name"])
    print("Amount:", highest_expense["amount"])

final_report()
```

```python
#Task 12: Store Inventory Manager
inventory = {}


def add_product():
    name = input("Enter product name: ").title()
    price = float(input("Enter price: "))
    quantity = int(input("Enter quantity: "))

    inventory[name] = {
        "price": price,
        "quantity": quantity
    }

    print("Product added successfully!")


def sell_product():
    name = input("Enter product name: ").title()

    if name in inventory:
        quantity = int(input("Enter quantity to sell: "))

        if inventory[name]["quantity"] >= quantity:
            inventory[name]["quantity"] -= quantity
            print("Product sold successfully!")

            if inventory[name]["quantity"] < 5:
                print("Warning: Low Stock!")
        else:
            print("Insufficient stock!")

    else:
        print("Product not found!")


def restock_product():
    name = input("Enter product name: ").title()

    if name in inventory:
        quantity = int(input("Enter quantity to add: "))
        inventory[name]["quantity"] += quantity
        print("Stock updated successfully!")

    else:
        print("Product not found!")


def search_product():
    name = input("Enter product name: ").title()

    if name in inventory:
        print("Product:", name)
        print("Price:", inventory[name]["price"])
        print("Quantity:", inventory[name]["quantity"])
    else:
        print("Product not found!")


def display_inventory():
    if len(inventory) == 0:
        print("Inventory is empty!")
        return

    total_value = 0

    print("\n--- Inventory ---")

    for name, details in inventory.items():
        price = details["price"]
        quantity = details["quantity"]

        print(name, "Price:", price, "Quantity:", quantity)

        if quantity < 5:
            print("  Low Stock!")

        total_value += price * quantity

    print("Total Inventory Value:", total_value)


while True:

    print("\n--- Store Inventory Manager ---")
    print("1. Add Product")
    print("2. Sell Product")
    print("3. Restock Product")
    print("4. Search Product")
    print("5. Display Inventory")
    print("6. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        add_product()

    elif choice == "2":
        sell_product()

    elif choice == "3":
        restock_product()

    elif choice == "4":
        search_product()

    elif choice == "5":
        display_inventory()

    elif choice == "6":
        print("Program ended.")
        break

    else:
        print("Invalid choice!")
```

```python
#Library book management
books = []


def add_book():
    book_id = input("Enter Book ID: ")

    for book in books:
        if book["id"] == book_id:
            print("Book ID already exists!")
            return

    title = input("Enter Book Title: ").title()
    author = input("Enter Author Name: ").title()

    book = {
        "id": book_id,
        "title": title,
        "author": author,
        "status": "Available"
    }

    books.append(book)
    print("Book added successfully!")


def display_books():
    if len(books) == 0:
        print("No books available!")
        return

    for book in books:
        print("----------------")
        print("Book ID:", book["id"])
        print("Title:", book["title"])
        print("Author:", book["author"])
        print("Status:", book["status"])


def search_book():
    title = input("Enter book title: ").title()

    for book in books:
        if book["title"] == title:
            print("Book ID:", book["id"])
            print("Title:", book["title"])
            print("Author:", book["author"])
            print("Status:", book["status"])
            return

    print("Book not found!")


def borrow_book():
    book_id = input("Enter Book ID: ")

    for book in books:
        if book["id"] == book_id:

            if book["status"] == "Available":
                book["status"] = "Borrowed"
                print("Book borrowed successfully!")
            else:
                print("Book is already borrowed!")

            return

    print("Book not found!")


def return_book():
    book_id = input("Enter Book ID: ")

    for book in books:
        if book["id"] == book_id:

            if book["status"] == "Borrowed":
                book["status"] = "Available"
                print("Book returned successfully!")
            else:
                print("Book is already available!")

            return

    print("Book not found!")


while True:

    print("\n--- Library Book Management System ---")
    print("1. Add Book")
    print("2. Display Books")
    print("3. Search Book")
    print("4. Borrow Book")
    print("5. Return Book")
    print("6. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        add_book()
    elif choice == "2":
        display_books()
    elif choice == "3":
        search_book()
    elif choice == "4":
        borrow_book()
    elif choice == "5":
        return_book()
    elif choice == "6":
        print("Program ended.")
        break
    else:
        print("Invalid choice!")
```

```python
#Resturant ordering system
menu={"Burger":500,"Pizza":1200,"Pasta":800,"Drink":200}
orders=[]
print("========Resturant menu===========")
for key,value in menu.items():
  print(f'{key}:{value}')
while True:
  item_name=input("Enter your item name:").title()
  if item_name in menu:
    quantity=int(input("Enter quantity of your item:"))
    price=menu[item_name]
    sub_total=price*quantity
    orders.append({"item":item_name,"quantity":quantity,"sub_total":sub_total})
  else:
    print("this item is not available")
  choice=input("you want to add more items:yes/no:")
  if choice=="no":
    break
total=0
for order in orders:
  total=total+order["sub_total"]
tax=total*0.13
if total>3000:
  discount=total*0.1
else:
  discount=0
grand_total=total+tax-discount
def receipt():
  print("==========Receipt=========")
  for order in orders:
    print(f'{order["item"]} x {order["quantity"]}={order["sub_total"]}')
  print("Total:",total)
  print("tax:",tax)
  print("Discount:",discount)
  print("Grand Total:",grand_total)
receipt()
```

```python

# Task 15: Clinic Appointment Manager
patients = []

def add_patient():
  patient_id = int(input("Enter your id: "))
  for patient in patients:
    if patient["id"] == patient_id:
      print("Patient already exists")
      return
  patient_name = input("Enter patient name: ")
  patient_age = int(input("Enter patient age: "))
  problem = input("Enter patient problem: ")
  time = int(input("Enter appointment time: "))
  emergency = input("Is emergency pateint? yes/no: ")

  patients.append({
    "id": patient_id,
    "name": patient_name,
    "age": patient_age,
    "problem": problem,
    "time": time,
    "emergency": emergency
  })

def search_patient():
  patient_id = int(input("Enter patient id: "))
  for patient in patients:
    if patient["id"] == patient_id:
      print("Patient found")
      print(f'Patient name: {patient["name"]}')
      print(f'Patient age: {patient["age"]}')
      print(f'Patient problem: {patient["problem"]}')
      print(f'Patient time: {patient["time"]}')
      print(f'Patient emergency: {patient["emergency"]}')
      return
    else:
      print("Patient not found")

def display_appointment():
  print("========== Emergency pateint ==========")
  for patient in patients:
    if patient["emergency"] == "yes":
      print(f'Patient name: {patient["name"]}')
      print(f'Patient age: {patient["age"]}')
      print(f'Patient problem: {patient["problem"]}')
      print(f'Patient time: {patient["time"]}')

  print("========== normal pateint ==========")
  for patient in patients:
    if patient["emergency"] == "no":
      print(f'Patient name: {patient["name"]}')
      print(f'Patient age: {patient["age"]}')
      print(f'Patient problem: {patient["problem"]}')
      print(f'Patient time: {patient["time"]}')

def delete_patient():
  patient_id = int(input("Enter patient id: "))
  for patient in patients:
    if patient["id"] == patient_id:
      patients.remove(patient)
      print("Patient deleted")

def count_patient():
  emergency_count = 0
  normal_count = 0
  for patient in patients:
    if patient["emergency"] == "yes":
      emergency_count += 1
    else:
      normal_count += 1

while True:
  print("1. Add patient")
  print("2. Search patient")
  print("3. Display appointment")
  print("4. Delete patient")
  print("5. Count patient")
  print("6. Exit")

  choice = int(input("Enter your choice: "))

  if choice == 1:
    add_patient()
  elif choice == 2:
    search_patient()
  elif choice == 3:
    display_appointment()
  elif choice == 4:
    delete_patient()
  elif choice == 5:
    count_patient()
  elif choice == 6:
    break
  else:
    print("Invalid choice")
```
