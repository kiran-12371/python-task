```python
#Simple Calculator
num1=float(input("enter a number:"))
num2=float(input("enter a sec number:"))
operator=input("enter a operator:(+,-,*,/)")
if operator=="+":
  print(num1+num2)
elif operator=="-":
  print(num1-num2)
elif operator=="*":
  print(num1*num2)
elif operator=="/":
  print(num1/num2)
else:
  print("INVALID OPERATOR")
```
