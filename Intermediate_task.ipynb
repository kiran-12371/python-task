```python
#intermideate level
#task 6
#grocery shopping bill
total_products=int(input("Enter the total products:"))
total_bill=0
for i in range(1,total_products+1):
  product_name=input(f"Enter the name of product{i}:")
  price=int(input(f"Enter the price of {product_name}:"))
  quantity=int(input(f"Enter the quantity of {product_name}:"))
  subtotal=price*quantity
  total_bill+=subtotal
if total_bill>=5000:
  discount=total_bill*10/100
  final_bill=total_bill-discount
else:
  discount=0
  final_bill=total_bill

print("=========/nGrocery Shopping Bil/n=======")
print(f"Total bill:{total_bill}")
print(f"Discount:{discount}")
print(f"Final bill:{final_bill}")
```

```python
#Employee Weekly salary
name=input("Enter your name:")
total_hours=int(input("Enter total working hour:"))
hour_rate=float(input("Enter hour rate:"))
if total_hours<=40:
 normal_salary=hour_rate*total_hours
 overtime_salary=0
else:
  normal_salary=hour_rate*40
  extra_hours=total_hours-40
  overtime_salary=extra_hours*hour_rate*1.5
total_salary=normal_salary+overtime_salary
print("-------/nTotal Weekly salary/n------")
print("Employee name:",name)
print("Normal salary:",normal_salary)
print("Overtime salary:",overtime_salary)
print(f"Total weekly salary:{total_salary}")
```

```python
# Message Analyze
message=input("Enter a message:")
print(message)
total_characters=len(message)
total_words=len(message.split())
vowels="aeiouAEIOU"
total_vowels=0
for char in message:
  if char in vowels:
    total_vowels=total_vowels+1
total_spaces=message.count(" ")
uppercase_msg=message.upper()
if "python"in message:
  python_found="yes"
else:
  python_found="no"
print("~~~~~~~/n Message Analyzer /n~~~~~~~")
print("uppercase message:",uppercase_msg)
print("Characters:",total_characters)
print("Words:",total_words)
print("vowels:",total_vowels)
print("spaces:",total_spaces)
print("python found:",python_found)
```

```python
#Contact Book
contacts={"ali":"031123245","sara":"0321345677"}
print("====contact book====")
print(contacts)
while True:
  print("1:Add Contact")
  print("2:Search contact")
  print("3:Display all contacts")
  print("4:Exit")
  choice=input("Enter your choice:")
  if choice=="1":
    name=input("enter your contact name:")
    if name in contacts:
      print("contact already exist:")
    else:
       phone=input("Enter phone number:")
       contacts[name]=phone
       print("contact added successfully:")
  elif choice=="2":
    name=input("enter name to serach contact:")
    if name in contacts:
       print("phone number:",contacts[name])
    else:
      print("contact not found.")
  elif choice=="3":
    if len(contacts)==0:
      print("contact not found.")
    else:
      print("All contacts:")
      for name,phone in contacts.items():
        print(name,"==",phone)
  elif choice=="4":
      print("program end.")
      break
  else:
      print("invalid choice.")

```
