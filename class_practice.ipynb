```python
#Cinema Ticket Price
name=input("Enter your name:")
age=int(input("Enter your age:"))
if age<5 or age>100:
  print("Free tickets")
elif age==5 or age<=12:
   price=300
elif age==13 or age<=59:
   price=700
elif age==60 or age<=100:
   price=400
else:
  print("invalid input")

print("Customer name:",name)
print("Ticket price:",price)
```

```python
#Temperature Converter
celsius=int(input("Enter celsius temperature:"))
fahrenheit=(celsius*9/5)+32
if celsius<10 and celsius==0:
  temperature="very cold"
elif celsius>=10 and celsius<=19:
  temperature="cold"
elif celsius>=20 and celsius<=29:
  temperature="normal"
else:
  temperature="hot"
print('fahrenheit is:',fahrenheit)
print('weather condition is:',temperature)
```

```python

# weekly steps
for i in range(1,8):
  per_day_steps = int(input("Enter the number of steps:"))
  total_steps = per_day_steps*7
  daily_average = total_steps/7

if  daily_average >= 5000:
    print("active week")
else:
    print("needs more activity")
print("Total steps:",total_steps)
print("Daily average:",daily_average)
```
